```python
import os
import sys

import pytest

VOLUME = "/Volumes/avaliacao_sicredi/novarota/dados"
os.environ["PIPELINE_DATA_ROOT"] = VOLUME

sys.dont_write_bytecode = True

TEMPORARIOS = f"{VOLUME}/_pytest"
os.makedirs(TEMPORARIOS, exist_ok=True)
os.environ["PYTEST_DEBUG_TEMPROOT"] = TEMPORARIOS

REPO_ROOT = os.path.dirname(os.getcwd())
if REPO_ROOT not in sys.path:
    sys.path.insert(0, REPO_ROOT)
os.chdir(REPO_ROOT)

print(f"raiz do projeto ....: {REPO_ROOT}")
print(f"raiz de dados ......: {VOLUME}")
print(f"temporarios pytest .: {TEMPORARIOS}")
```

```python
# --assert=plain        desativa a reescrita de asserts, que exige escrever .pyc
# -p no:cacheprovider   evita a criacao do diretorio .pytest_cache
retcode = pytest.main(["tests", "-v", "--assert=plain", "-p", "no:cacheprovider"])
# Falha a celula quando houver teste quebrado
assert retcode == 0, "A suite de testes falhou. Ver log acima."
```

```python
# Evidência da execução
import contextlib

caminho = f"{VOLUME}/suite_databricks.txt"

with open(caminho, "w") as arquivo:
    with contextlib.redirect_stdout(arquivo):
        pytest.main(["tests", "-v", "--assert=plain", "-p", "no:cacheprovider"])

with open(caminho) as arquivo:
    print(arquivo.read())
```
